### 解決 403 錯誤：加入 User-Agent
透過 `storage_options` 參數，我們可以自訂 HTTP 標頭（Headers），讓網站認為這是一般瀏覽器發出的請求。

```python
import pandas as pd

url = 'https://en.wikipedia.org/wiki/List_of_countries_and_dependencies_by_population'

headers = {'User-Agent': 'Mozilla/5.0 (Windows NT 10.0; Win64; x64) AppleWebKit/537.36 (KHTML, like Gecko) Chrome/91.0.4472.124 Safari/537.36'}
tables = pd.read_html(url, storage_options=headers)

print(len(tables))

df = tables[0]
display(df.head())
```

```python
import numpy as np
import pandas as pd

# Adding more common angles
degrees = [0, 30, 45, 60, 90, 180, 270, 360]
sine_values = np.sin(np.deg2rad(degrees))

# Clean up floating point precision issues (e.g. sin(180) should be 0)
sine_values = np.round(sine_values, 10)

sine_table_expanded = pd.Series(sine_values, index=degrees)
sine_table_expanded
```

```python
import pandas as pd

# DataFrame 1: Name and Age
df_age = pd.DataFrame({
    'Name': ['Alice', 'Bob', 'Charlie', 'Alice'],
    'Age': [25, 30, 35, 40]
})

# DataFrame 2: Name and Height
df_height = pd.DataFrame({
    'Name': ['Bob', 'Charlie','Alice',  'Alice'],
    'Height': [165, 180, 175, 160]
})

print("--- DataFrame 1: Age ---")
display(df_age)
print("\n--- DataFrame 2: Height ---")
display(df_height)
```

```python
pd.merge(df_age, df_height, on='Name', how='left')
```

```python
import pandas as pd

# DataFrame 1: Name and Age
df_age = pd.DataFrame({
    'Name': ['Alice', 'Bob', 'Charlie', 'Davide'],
    'Age': [25, 30, 35, 40]
})
df_age.set_index('Name', inplace=True)
# DataFrame 2: Name and Height
df_height = pd.DataFrame({
    'Name': ['Bob', 'Charlie','Eve',  'John'],
    'Height': [165, 180, 175, 160]
})
df_height.set_index('Name', inplace=True)

print("--- DataFrame 1: Age ---")
display(df_age)
print("\n--- DataFrame 2: Height ---")
display(df_height)
```

```python
pd.concat([df_age, df_height])
```
