### 解決 403 錯誤：加入 User-Agent
透過 `storage_options` 參數，我們可以自訂 HTTP 標頭（Headers），讓網站認為這是一般瀏覽器發出的請求。

```python
import pandas as pd

url = 'https://en.wikipedia.org/wiki/List_of_countries_and_dependencies_by_population'

headers = {'User-Agent': 'Mozilla/5.0 (Windows NT 10.0; Win64; x64) AppleWebKit/537.36 (KHTML, like Gecko) Chrome/91.0.4472.124 Safari/537.36'}
tables = pd.read_html(url, storage_options=headers)

print(len(tables))

df = tables[0]
display(df.head())
```

```python
import numpy as np
import pandas as pd

degrees = [30, 45, 90]
sine_values = np.sin(np.deg2rad(degrees))

sine_table = pd.Series(sine_values, index=degrees)
sine_table
```

```python
import numpy as np
import pandas as pd

# Adding more common angles
degrees = [0, 30, 45, 60, 90, 180, 270, 360]
sine_values = np.sin(np.deg2rad(degrees))

# Clean up floating point precision issues (e.g. sin(180) should be 0)
sine_values = np.round(sine_values, 10)

sine_table_expanded = pd.Series(sine_values, index=degrees)
sine_table_expanded
```

```python
import pandas as pd

# Create a simple dictionary
data = {
    'Name': ['Alice', 'Bob', 'Charlie'],
    'Age': [25, 30, 35],
    'City': ['New York', 'London', 'Paris']
}

# Construct DataFrame using from_dict
df_dict = pd.DataFrame.from_dict(data)
df_dict
```

```python
import pandas as pd

# Create a dictionary where we want the keys to be rows
data = {
    'Row_A': [1, 'a', 3],
    'Row_B': [4, 'b', 6],
    'Row_C': [7, 'c', 9]
}

# Construct DataFrame using from_dict with orient='index'
df_orient = pd.DataFrame.from_dict(data, orient='index', columns=['Col_1', 'Col_2', 'Col_3'])
df_orient
```

```python
import pandas as pd

# Create a list of tuples representing records
records = [
    (1, 'Alice', 'Engineer'),
    (2, 'Bob', 'Designer'),
    (3, 'Charlie', 'Artist')
]

# Construct DataFrame using from_records and specify column names
df_records = pd.DataFrame.from_records(records, columns=['ID', 'Name', 'Role'])
df_records
```
