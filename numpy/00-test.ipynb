# NumPy 基礎觀念教學

NumPy (Numerical Python) 是 Python 進行資料科學與數值運算的基礎套件。它提供了高效能的多維陣列物件 (ndarray) 以及操作這些陣列的工具。

我們先從引入 NumPy 套件開始：

```python
import numpy as np

# 確認 NumPy 版本
print(f"NumPy version: {np.__version__}")
```

## 1. 建立 NumPy 陣列 (Arrays)

建立陣列最基本的方法是使用 `np.array()`，將 Python 的列表 (List) 轉換為 NumPy 陣列。此外，NumPy 也提供了許多方便的函數來建立特定內容的陣列，例如全零、全一或等差數列。

```python
# 1. 從列表建立一維陣列 (1D Array)
a = np.array([1, 2, 3, 4, 5])
print("一維陣列 a:", a)

# 2. 建立全零陣列 (例如 2列3行的矩陣)
zeros_arr = np.zeros((2, 3))
print("\n2x3 全零陣列 zeros_arr:\n", zeros_arr)

# 3. 建立全一陣列
ones_arr = np.ones((2, 2))
print("\n2x2 全一陣列 ones_arr:\n", ones_arr)

# 4. 使用 arange 建立等差陣列 (類似 Python 的 range)
# 參數: np.arange(起始, 結束(不包含), 步長)
seq_arr = np.arange(0, 10, 2)
print("\n等差陣列 seq_arr:", seq_arr)

# 5. 使用 linspace 建立指定區間內均勻分布的數值
# 參數: np.linspace(起始, 結束(包含), 總數量)
lin_arr = np.linspace(0, 1, 5)
print("\n均勻分布陣列 lin_arr:", lin_arr)
```

## 2. 陣列的基本屬性 (Attributes)

了解 NumPy 陣列的形狀與型別非常重要。以下是幾個最常用的屬性：
- `ndim`: 陣列的維度 (例如：1維、2維)。
- `shape`: 陣列的形狀 (各維度的大小，以 Tuple 表示)。
- `size`: 陣列中元素的總數。
- `dtype`: 陣列中元素的資料型別。

```python
# 建立一個二維陣列 (矩陣)
matrix = np.array([
    [1.5, 2.2, 3.1],
    [4.8, 5.0, 6.9]
])

print("矩陣 matrix:\n", matrix)
print("----------------------")
print("維度 (ndim):", matrix.ndim)
print("形狀 (shape):", matrix.shape) # 回傳 (2, 3) 代表 2列 3行
print("元素總數 (size):", matrix.size)
print("資料型別 (dtype):", matrix.dtype)
```

## 3. 陣列視覺化 (Visualization)

結合 `matplotlib` 套件，我們可以很方便地將 NumPy 陣列的數值繪製成圖表。

```python
import matplotlib.pyplot as plt

# 視覺化一維陣列 (折線圖)
plt.figure(figsize=(8, 4))
plt.plot(a, marker='o')
plt.title('Visualization of 1D Array `a`')
plt.xlabel('Index')
plt.ylabel('Value')
plt.grid(True)
plt.show()
```

```python
# 視覺化二維陣列 (熱力圖)
plt.figure(figsize=(6, 4))
plt.imshow(matrix, cmap='viridis')
plt.colorbar(label='Value')
plt.title('Visualization of 2D Matrix')
plt.xlabel('Column Index')
plt.ylabel('Row Index')
plt.show()
```

```python
# 更多視覺化方式
import matplotlib.pyplot as plt
import numpy as np

# 1. 一維陣列 - 長條圖 (Bar Chart)
plt.figure(figsize=(8, 4))
plt.bar(range(len(a)), a, color='skyblue', edgecolor='black')
plt.title('Bar Chart of 1D Array `a`')
plt.xlabel('Index')
plt.ylabel('Value')
plt.xticks(range(len(a)))
plt.show()

# 2. 二維陣列 - 3D 表面圖 (3D Surface Plot)
fig = plt.figure(figsize=(8, 6))
ax = fig.add_subplot(111, projection='3d')

# 建立 X, Y 網格
x = np.arange(matrix.shape[1])
y = np.arange(matrix.shape[0])
X, Y = np.meshgrid(x, y)

# 繪製 3D 表面
surf = ax.plot_surface(X, Y, matrix, cmap='plasma')
fig.colorbar(surf, label='Value')
ax.set_title('3D Surface Plot of 2D Matrix')
ax.set_xlabel('Column Index')
ax.set_ylabel('Row Index')
ax.set_zlabel('Value')

plt.show()
```
